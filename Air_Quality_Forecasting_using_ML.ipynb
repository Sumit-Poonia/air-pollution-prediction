```python
# training the Prophet model
model = Prophet()
model.fit(data)
```

```python
future = model.make_future_dataframe(periods=365, freq = 'H')
future.tail()
```

```python
data = {
    'Gas': [
        'CO(GT)', 'PT08.S1(CO)', 'NMHC(GT)', 'C6H6(GT)', 'PT08.S2(NMHC)',
        'NOx(GT)', 'PT08.S3(NOx)', 'NO2(GT)', 'PT08.S4(NO2)', 'PT08.S5(O3)',
        'T', 'RH', 'AH'
    ],
    'Good': [
        '0–2.0 mg/m³', '0–1000', '0–200 µg/m³', '0–5 µg/m³', '0–1000',
        '0–53 ppb', '0–1000', '0–40 µg/m³', '0–1000', '0–1000',
        '10–30°C', '30–60%', '5–10'
    ],
    'Average': [
        '2.1–10.0 mg/m³', '1001–2000', '201–1000 µg/m³', '5.1–15 µg/m³', '1001–2000',
        '54–100 ppb', '1001–2000', '41–80 µg/m³', '1001–2000', '1001–2000',
        '<10 or 31–35°C', '20–29% or 61–80%', '2.5–4.9 or 10.1–15'
    ],
    'Poor': [
        '>10.0 mg/m³', '>2000', '>1000 µg/m³', '>15 µg/m³', '>2000',
        '>100 ppb', '>2000', '>80 µg/m³', '>2000', '>2000',
        '<5 or >35°C', '<20% or >80%', '<2.5 or >15'
    ]
}

df = pd.DataFrame(data)

# Print the table
print(df)

# Save to CSV
df.to_csv("gas_thresholds.csv", index=False)
```

```python
forecast = model.predict(future)

thresholds_df = pd.read_csv("gas_thresholds.csv")
# For the gas for we want to predict
gas_name = 'NO2(GT)'
row = thresholds_df[thresholds_df['Gas'] == gas_name].iloc[0]

good_range = row['Good'].split('–')
avg_range = row['Average'].split('–')
poor_min = float(row['Poor'].replace('>','').split()[0])

good_min = float(good_range[0])
good_max = float(good_range[1].split()[0])
avg_min = float(avg_range[0])
avg_max = float(avg_range[1].split()[0])

def categorize_dynamic(value):
    if good_min <= value <= good_max:
        return 'good'
    elif avg_min <= value <= avg_max:
        return 'average'
    elif value > poor_min:
        return 'danger'
    else:
        return 'unknown'

forecast['category'] = forecast['yhat'].apply(categorize_dynamic)

print(forecast[['ds','yhat','yhat_lower','yhat_upper','category']].head(20))
```
